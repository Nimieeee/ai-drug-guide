The notebook from the video "How to turn a molecule into data". It takes huperzine A, the acetylcholinesterase inhibitor, and turns it into an ECFP4 fingerprint with RDKit: 2,048 positions, 41 of them set.

Run the cells from top to bottom. The first cell installs RDKit on Colab if it is missing.

```python
# --- SETUP: runs on Colab, Kaggle or on your own computer -------------------------------
import sys, importlib.util
ON_COLAB = 'google.colab' in sys.modules
if ON_COLAB and importlib.util.find_spec('rdkit') is None:
    print('Installing RDKit (about a minute) ...')
    !pip install -q rdkit
```

```python
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator
import numpy as np

# huperzine A, the acetylcholinesterase inhibitor from the reel
smi = "C/C=C/1\\[C@@H]2CC3=C([C@]1(CC(=C2)C)N)C=CC(=O)N3"
mol = Chem.MolFromSmiles(smi)
print(mol.GetNumAtoms(), "heavy atoms")
```

```python
gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)
fp = gen.GetFingerprint(mol)
bits = np.array(fp)
print(len(bits), int(bits.sum()))
```

```python
np.flatnonzero(bits)[:10]
```

```python
counts = gen.GetCountFingerprint(mol).GetNonzeroElements()
print(len(counts), max(counts.values()))
```

## What the numbers mean

- **2,048** is the length of the vector: one position for every possible substructure hash.
- **41** positions are 1. The rest are 0.
- The first ten set positions are 27, 81, 84, 231, 291, 314, 338, 542, 650, 676.
- **5** is the highest count in the count fingerprint: one substructure occurs five times.

Two different substructures can land on the same position. That is a hash collision. A larger fingerprint gives them less room. A count-based fingerprint does not stop the collision, but it records how many times each position was hit, so the information is not flattened to a single 1.
